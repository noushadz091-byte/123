## Question 23 — Nested Salary Calculator

**Problem:**  
Salary calculation has several smaller steps. Use nested functions to organize the logic cleanly.

**Requirements:**
- Create an outer function called `calculate_salary()`.

- The outer function should combine the results and return the final salary.
- Call the outer function and display the result.

**Focus:** Nested functions

````python
# Write your code below
def calculate_salary(basic_salary):
    def calculate_allowance():
        return basic_salary * 0.10

    def calculate_deduction():
        return basic_salary * 0.05

    def final_salary():
        return basic_salary + calculate_allowance() - calculate_deduction()

    return final_salary()

print("Final Salary:", calculate_salary(50000))


````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
# Write your code below
def process_order(price, quantity):
    def calculate_subtotal():
        return price * quantity

    def calculate_discount():
        return calculate_subtotal() * 0.10

    def calculate_tax():
        return calculate_subtotal() * 0.05

    def calculate_final_amount():
        return calculate_subtotal() - calculate_discount() + calculate_tax()

    return calculate_final_amount()

print("Final Order Amount:", process_order(1000, 3))
````

Design the program yourself. Decide:

- What each function should return

This question tests your **complete understanding** of Python functions.

**Focus:** Complete modular real-world system

````python
# Write your code below
def customer_info(name, city):
    return name, city

def product_total(price, quantity):
    return price * quantity

def calculate_subtotal(*items):
    return sum(items)

def calculate_discount(subtotal, discount_percent=10):
    return subtotal * discount_percent / 100

def calculate_tax(amount):
    return amount * 0.05

def delivery_charges(subtotal):
    return 0 if subtotal >= 5000 else 200

def calculate_final_bill(subtotal):
    def final_calculation():
        discount = calculate_discount(subtotal)
        amount_after_discount = subtotal - discount
        tax = calculate_tax(amount_after_discount)
        delivery = delivery_charges(subtotal)
        return amount_after_discount + tax + delivery
    return final_calculation()

name, city = customer_info("Ali", "Karachi")
item1 = product_total(1500, 2)
item2 = product_total(1000, 1)
subtotal = calculate_subtotal(item1, item2)
format_price = lambda amount: f"Rs. {amount:.2f}"

print("========== ORDER FUNCTION HAS MADE ==========")
print("Customer:", name)
print("City:", city)
print("Subtotal:", format_price(subtotal))
print("Discount:", format_price(calculate_discount(subtotal)))
print("Final Bill:", format_price(calculate_final_bill(subtotal)))
````
